```python
VALID_AA = set("ACDEFGHIKLMNPQRSTVWY")

def invalid_chars(sequence):
    return set(sequence) - VALID_AA
```

```python
df["invalid_chars"] = df["md5"].apply(invalid_chars)
df["invalid_chars"]
```
